## Noise based on Various Distribution



```python
import cv2
import numpy as np 
from matplotlib import pyplot as plt
```

```python
img_path = '../../images/cameraman.png'
```

## Gaussian Noise

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt

# Read the grayscale image
img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)

# Generate Gaussian noise
mean = 0
stddev = 60  # Adjust standard deviation for noise intensity
gaussian_noise = np.random.normal(mean, stddev, img.shape)

# Add Gaussian noise to the image
noisy_img = img + gaussian_noise

# Clip the pixel values to stay within valid range [0, 255]
noisy_img = np.clip(noisy_img, 0, 255).astype(np.uint8)

# Display results
plt.figure(figsize=(20, 12))

# Display the original image
plt.subplot(2, 2, 1)
plt.title("Original Image")
plt.imshow(img, cmap='gray')
plt.axis('off')

# Display the noisy image
plt.subplot(2, 2, 2)
plt.title("Gaussian Noisy Image")
plt.imshow(noisy_img, cmap='gray')
plt.axis('off')

# Display histogram of the original image
plt.subplot(2, 2, 3)
plt.title("Histogram of Original Image")
plt.hist(img.ravel(), bins=256, range=[0, 256], color='blue', alpha=0.7)
plt.xlabel('Pixel Intensity')
plt.ylabel('Frequency')

# Display histogram of the noisy image
plt.subplot(2, 2, 4)
plt.title("Histogram of Noisy Image")
plt.hist(noisy_img.ravel(), bins=256, range=[0, 256], color='red', alpha=0.7)
plt.xlabel('Pixel Intensity')
plt.ylabel('Frequency')


plt.tight_layout()
plt.show()
```

## Salt-and-Pepper Noise

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt

def add_salt_pepper_noise(img, salt_prob, pepper_prob):
    """
    Add salt and pepper noise to an image.
    :param img: Input grayscale image.
    :param salt_prob: Probability of salt noise.
    :param pepper_prob: Probability of pepper noise.
    :return: Noisy image.
    """
    noisy_img = img.copy()
    total_pixels = img.size
    
    if salt_prob > 0:

        # Salt noise
        num_salt = int(total_pixels * salt_prob)
        coords_salt = [np.random.randint(0, i - 1, num_salt) for i in img.shape]
        noisy_img[coords_salt[0], coords_salt[1]] = 255

    if pepper_prob > 0:
        
        # Pepper noise
        num_pepper = int(total_pixels * pepper_prob)
        coords_pepper = [np.random.randint(0, i - 1, num_pepper) for i in img.shape]
        noisy_img[coords_pepper[0], coords_pepper[1]] = 0

    return noisy_img

# Read the grayscale image
img = cv2.imread('../../Image Processing Labs/images/apple.jpg', cv2.IMREAD_GRAYSCALE)

# Add Salt, Pepper, and Salt-Pepper noise
salt_prob = 0.01 
pepper_prob = 0.01  
salt_pepper_noisy_img = add_salt_pepper_noise(img, salt_prob, pepper_prob)


# Display results
plt.figure(figsize=(20, 12))
plt.subplot(1, 2, 1)
plt.title("Original Image")
plt.imshow(img, cmap='gray')

plt.subplot(1, 2, 2)
plt.title("Salt-Pepper Noisy Image")
plt.imshow(salt_pepper_noisy_img, cmap='gray')

plt.show()
```

## Rayleigh Noise

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt

def add_rayleigh_noise(img, scale):
    """
    Add Rayleigh noise to an image.
    :param img: Input grayscale image.
    :param scale: Scale parameter for the Rayleigh distribution.
    :return: Noisy image.
    """
    # Generate Rayleigh noise
    rayleigh_noise = np.random.rayleigh(scale, img.shape)
    
    # Add the noise to the image
    noisy_img = img + rayleigh_noise
    
    # Clip the pixel values to stay in valid range [0, 255]
    noisy_img = np.clip(noisy_img, 0, 255).astype(np.uint8)
    
    return noisy_img

# Read the grayscale image
img = cv2.imread('../../Image Processing Labs/images/apple.jpg', cv2.IMREAD_GRAYSCALE)

# Add Rayleigh noise
scale = 55  # Adjust the scale parameter to control noise intensity
rayleigh_noisy_img = add_rayleigh_noise(img, scale)

# Display results
plt.figure(figsize=(20, 12))
plt.subplot(1, 2, 1)
plt.title("Original Image")
plt.imshow(img, cmap='gray')

plt.subplot(1, 2, 2)
plt.title("Rayleigh Noisy Image")
plt.imshow(rayleigh_noisy_img, cmap='gray')

plt.show()
```

## Erlang (Gamma) Noise

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt

def add_gamma_noise(img, shape, scale):
    """
    Add Gamma (Erlang) noise to an image.
    :param img: Input grayscale image.
    :param shape: Shape parameter (k) for the Gamma distribution.
    :param scale: Scale parameter (θ) for the Gamma distribution.
    :return: Noisy image.
    """
    # Generate Gamma noise
    gamma_noise = np.random.gamma(shape, scale, img.shape)
    
    # Add the noise to the image
    noisy_img = img + gamma_noise
    
    # Clip the pixel values to stay in valid range [0, 255]
    noisy_img = np.clip(noisy_img, 0, 255).astype(np.uint8)
    
    return noisy_img

# Read the grayscale image
img = cv2.imread('../../Image Processing Labs/images/apple.jpg', cv2.IMREAD_GRAYSCALE)

# Add Gamma (Erlang) noise
shape = 2   # Shape parameter (k)
scale = 55  # Scale parameter (θ)
gamma_noisy_img = add_gamma_noise(img, shape, scale)

# Display results
plt.figure(figsize=(20, 12))
plt.subplot(1, 2, 1)
plt.title("Original Image")
plt.imshow(img, cmap='gray')

plt.subplot(1, 2, 2)
plt.title("Gamma (Erlang) Noisy Image")
plt.imshow(gamma_noisy_img, cmap='gray')

plt.show()
```

## Exponential Noise

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt

def add_exponential_noise(img, scale):
    """
    Add Exponential noise to an image.
    :param img: Input grayscale image.
    :param scale: Scale parameter (λ^-1) for the Exponential distribution.
    :return: Noisy image.
    """
    # Generate Exponential noise
    exponential_noise = np.random.exponential(scale, img.shape)
    
    # Add the noise to the image
    noisy_img = img + exponential_noise
    
    # Clip the pixel values to stay in valid range [0, 255]
    noisy_img = np.clip(noisy_img, 0, 255).astype(np.uint8)
    
    return noisy_img

# Read the grayscale image
img = cv2.imread('../../Image Processing Labs/images/apple.jpg', cv2.IMREAD_GRAYSCALE)

# Add Exponential noise
scale = 55  # Scale parameter (λ^-1) controlling noise intensity
exponential_noisy_img = add_exponential_noise(img, scale)

# Display results
plt.figure(figsize=(20, 12))
plt.subplot(1, 2, 1)
plt.title("Original Image")
plt.imshow(img, cmap='gray')

plt.subplot(1, 2, 2)
plt.title("Exponential Noisy Image")
plt.imshow(exponential_noisy_img, cmap='gray')

plt.show()
```

## Uniform Noise

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt

def add_uniform_noise(img, low, high):
    """
    Add Uniform noise to an image.
    :param img: Input grayscale image.
    :param low: Lower bound of the uniform distribution.
    :param high: Upper bound of the uniform distribution.
    :return: Noisy image.
    """
    # Generate Uniform noise
    uniform_noise = np.random.uniform(low, high, img.shape)
    
    # Add the noise to the image
    noisy_img = img + uniform_noise
    
    # Clip the pixel values to stay in valid range [0, 255]
    noisy_img = np.clip(noisy_img, 0, 255).astype(np.uint8)
    
    return noisy_img

# Read the grayscale image
img = cv2.imread('../../Image Processing Labs/images/apple.jpg', cv2.IMREAD_GRAYSCALE)

# Add Uniform noise
low = -30  # Lower bound of noise
high = 80  # Upper bound of noise
uniform_noisy_img = add_uniform_noise(img, low, high)

# Display results
plt.figure(figsize=(20, 12))
plt.subplot(1, 2, 1)
plt.title("Original Image")
plt.imshow(img, cmap='gray')

plt.subplot(1, 2, 2)
plt.title("Uniform Noisy Image")
plt.imshow(uniform_noisy_img, cmap='gray')

plt.show()
```

## Noise with Historgrams

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt

def add_gaussian_noise(img, mean, stddev):
    gaussian_noise = np.random.normal(mean, stddev, img.shape)
    noisy_img = img + gaussian_noise
    return np.clip(noisy_img, 0, 255).astype(np.uint8)

def add_rayleigh_noise(img, scale):
    rayleigh_noise = np.random.rayleigh(scale, img.shape)
    noisy_img = img + rayleigh_noise
    return np.clip(noisy_img, 0, 255).astype(np.uint8)

def add_gamma_noise(img, shape, scale):
    gamma_noise = np.random.gamma(shape, scale, img.shape)
    noisy_img = img + gamma_noise
    return np.clip(noisy_img, 0, 255).astype(np.uint8)

def add_exponential_noise(img, scale):
    exponential_noise = np.random.exponential(scale, img.shape)
    noisy_img = img + exponential_noise
    return np.clip(noisy_img, 0, 255).astype(np.uint8)

def add_uniform_noise(img, low, high):
    uniform_noise = np.random.uniform(low, high, img.shape)
    noisy_img = img + uniform_noise
    return np.clip(noisy_img, 0, 255).astype(np.uint8)

# Read the grayscale image
img = cv2.imread('../../Image Processing Labs/images/shape1.png', cv2.IMREAD_GRAYSCALE)

# Apply different noise types
scale = 5
gaussian_noisy_img = add_gaussian_noise(img, mean=0, stddev=25)
rayleigh_noisy_img = add_rayleigh_noise(img, scale=scale)
gamma_noisy_img = add_gamma_noise(img, shape=2, scale=scale)
exp_noisy_img = add_exponential_noise(img, scale=scale)
uniform_noisy_img = add_uniform_noise(img, low=-30, high=30)

# Plot images and histograms
plt.figure(figsize=(20, 16))

noises = [
    ("Gaussian Noise", gaussian_noisy_img),
    ("Rayleigh Noise", rayleigh_noisy_img),
    ("Gamma Noise", gamma_noisy_img),
    ("Exponential Noise", exp_noisy_img),
    ("Uniform Noise", uniform_noisy_img),
]

for i, (title, noisy_image) in enumerate(noises):
    # Plot noisy image
    plt.subplot(5, 2, i * 2 + 1)
    plt.title(f"{title}")
    plt.imshow(noisy_image, cmap='gray')
    plt.axis('off')

    # Plot histogram of noisy image
    plt.subplot(5, 2, i * 2 + 2)
    plt.title(f"Histogram ({title})")
    plt.hist(noisy_image.ravel(), bins=256, range=[0, 256], color='gray')

plt.tight_layout()
plt.show()
```
